# make senior citizen column readble

```python
df["senior_citizen"] = df["senior_citizen"].map({0: "No", 1: "Yes"})
```

# Feature Engineering (New columns)

```python
# tenure_group
bins = [0, 12, 24, 48, 60, np.inf]
labels = ["0-12", "13-24", "25-48", "49-60", "60+"]
df["tenure_group"] = pd.cut(df["tenure"], bins=bins, labels=labels, right=True, include_lowest=True)
```

```python
# avg_monthly_spend
df["avg_monthly_spend"] = np.where(
    df["tenure"] > 0,
    df["total_charges"] / df["tenure"],
    df["monthly_charges"] 
)
```

```python
# num_services
service_cols = [
    "phone_service", "internet_service", "online_security", "online_backup",
    "device_protection", "tech_support", "streaming_tv", "streaming_movies"
]
df["num_services"] = df[service_cols].apply(
    lambda row: sum(row == "Yes"), axis=1
)
```

```python
# is_new_customer
df["is_new_customer"] = df["tenure"] < 6
```
